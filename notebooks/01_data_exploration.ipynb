# 1. Data exploration

Inspect one downloaded Ookla quarter and the companion ACS tract table. Run `python -m src.ingest.download_ookla --limit 1` and `python -m src.ingest.download_census` first.


```python
from pathlib import Path
import pandas as pd
import pyarrow.parquet as pq

root = Path('.')
ookla_files = sorted((root / 'data/ookla').rglob('*.parquet'))
census_file = root / 'data/census/acs5_tract_2022.parquet'
if not ookla_files:
    print('No Ookla Parquet file found. Run the ingestion command above.')
else:
    metadata = pq.ParquetFile(ookla_files[0]).metadata
    print('File:', ookla_files[0], 'rows:', metadata.num_rows)
    print(pq.read_schema(ookla_files[0]))
    sample = pd.read_parquet(ookla_files[0]).head(10)
    display(sample)
if census_file.exists():
    census = pd.read_parquet(census_file)
    display(census.head())
    display(census[['broadband_adoption_pct', 'median_income', 'education_bachelors_pct']].describe())
else:
    print(f'Missing {census_file}; run Census ingestion.')
```

## Notes

Ookla speed fields are in kilobits per second in the source and are converted to Mbps in Spark. ACS adoption is calculated as broadband-of-any-type households divided by total households; education is the share of adults 25+ with a bachelor's degree or higher.
