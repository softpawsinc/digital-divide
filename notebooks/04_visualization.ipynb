# 4. Serving and visualization

After running the pipeline, cluster rows are stored in MongoDB and joined to TIGER tract boundaries in a local GeoJSON file. The Streamlit app provides filters and interactive maps.


```python
from pathlib import Path
import pandas as pd

geojson_path = Path('data/processed/clusters.geojson')
if geojson_path.exists():
    import geopandas as gpd
    tracts = gpd.read_file(geojson_path)
    print(tracts.shape, tracts.crs)
    display(tracts.drop(columns='geometry').head())
else:
    print('Missing map export. Run python -m src.visualize.export_geojson after MongoDB is populated.')
```

```python
import subprocess

# Start the interactive app in a terminal from the project root.
print('streamlit run src/visualize/dashboard.py')
```

## Dashboard guide

Select state FIPS, segment, and household-income range in the sidebar. The map and tables respond to these filters. Summary panels show cluster size, speed, and adoption. MongoDB is expected at `localhost:27017` when the app runs on your laptop.
