# 3. Clustering and evaluation

Run K-Means over the six imputed tract features. StandardScaler normalizes the features, and candidate k values are compared with WSSSE and a deterministic sample silhouette score. The highest sampled silhouette selects the final k.


```python
import json
import subprocess
from pathlib import Path

command = 'cd /workspace && spark-submit --master local[*] /workspace/src/model/kmeans_clustering.py'
subprocess.run(['docker', 'compose', '-f', 'docker/docker-compose.yml', 'exec', '-T', 'spark', 'bash', '-lc', command], check=True)
metrics = json.loads(Path('data/processed/model_metrics.json').read_text())
print('Selected k:', metrics['selected_k'])
print(metrics['metrics'])
```

```python
from IPython.display import Image, display
plot = Path('data/processed/plots/kmeans_evaluation.png')
if plot.exists():
    display(Image(filename=str(plot)))
```
